````python
import pandas as pd
import geopandas as gpd
````

## Create `session.shp`

The `session.shp` file contains the following fields.

|Field Name|Required|Description|Data Type|
|----------|------|------------|---------|
|rc_id|Yes|Regional Center ID|string|
|analy_time|Yes|Date analysis done - could be year or year/month|string|
|src_date|No|Timestamp data used in mapping acquired|string|
|data_src|Yes|Source of data used for mapping|string|
|proc_desc|Yes|Short description of mapping method|string|
|anlst_surn|Yes|Surname of analyst|string|
|anlst_givn|Yes|Given names of Analyst|string|
|3d_desc|Yes|Description of how 3D information derived|string|

Currently, the input is a GMT (Generic Mapping Tools) format file.  Each analyst is documented in a single line.

Example GMT format file.
```
# @VGMT1.0 @GPOINT
# @R-123.7742/-123.1238/47.5231/47.9104
# @Jp"+proj=longlat +datum=WGS84 +no_defs "
# @Jw"GEOGCS[\"GCS_WGS_1984\",DATUM[\"WGS_1984\",SPHEROID[\"WGS_84\",6378137.0,298.257223563]],PRIMEM[\"Greenwich\",0.0],UNIT[\"Degree\",0.0174532925199433]]"
# @Nrc_id|analy_time|data_src|proc_desc|anlst_surn|anlst_givn|3d_desc
# @Tstring|string|string|string|string|string|string
# FEATURE_DATA
>
# @D507|2017-10-24|"NAIP imagery and LiDAR data"|"Manual digitization"|Mozer|Max|""
-123.4 47.6
>
# @D507|2017-10-24|"NAIP imagery and LiDAR data"|"Manual digitization"|Pflug|Justin|""
-123.4 47.6
```


````python
# Data
EPSG = "4326"

columns = [
    "rc_id",
    "analy_time",
    "data_src",
    "proc_desc",
    "anlst_surn",
    "anlst_givn",
    "3d_desc",
]

dtype = [
    str,
    str,
    str,
    str,
    str,
    str,
    str,
]

point = (-123.4, 47.6)
data = [
    {
        "rc_id": "507",
        "analy_time": "2017-10-24",
        "data_src": "NAIP imagery and LiDAR data",
        "proc_desc": "Manual digitization",
        "anlst_surn": "Mozer",
        "anlst_givn": "Max",
        "3d_desc": "",
    },
        {
        "rc_id": "507",
        "analy_time": "2017-10-24",
        "data_src": "NAIP imagery and LiDAR data",
        "proc_desc": "Manual digitization",
        "anlst_surn": "Pflug",
        "anlst_givn": "Justin",
        "3d_desc": "",
    }
]
````

````python
df = pd.DataFrame(data)
df
````

````python
x, y = zip(*[point, point])
gdf = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(x=x, y=y), crs=EPSG)
gdf
````

````python
gdf.to_file("test.shp")
````
